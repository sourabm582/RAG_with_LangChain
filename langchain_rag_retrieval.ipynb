**Retrieval & Generation**

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains import RetrievalQA
```

```python
#os.environ["GEMINI_API_KEY"] = "your_gemini_api_key"
```

```python
# configuration
vector_db_path = "D:/AI/RAG with LangChain/vector_db"
collection_name  = "document_collection"
```

```python
# loading the embedding model - default model
embedding = HuggingFaceEmbeddings()
```

```python
# initialize groq llm
import os

from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_classic.chains import RetrievalQA

# Initialize the Gemini model with your API key passed directly
llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",  # Note: ensure you use a valid current model name like gemini-1.5-flash or gemini-2.5-flash
    temperature=0.7,
    google_api_key="your_gemini_api_key"
)
```

```python
vector_store = Chroma(
    collection_name=collection_name,
    embedding_function=embedding,
    persist_directory=vector_db_path
)
```

```python
retriever = vector_store.as_retriever()
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    return_source_documents=True
)
```

```python
query = "What does the document say about Adaptive Radiation?"
response = qa_chain.invoke({"query":query})
print(response["result"])
print("-"*80)
for source in response["source_documents"]:
    print(source.metadata)
```
